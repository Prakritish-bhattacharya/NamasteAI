```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# 1. Sample documents (our "bags" of text)
documents = [
    "The cat sat on the mat.",
    "The dog sat on the rug.",
    "The cat and the dog played together.",
]

# 2. Initialize CountVectorizer
vectorizer = CountVectorizer()

# 3. Fit and transform the documents into a matrix of token counts
vector_matrix = vectorizer.fit_transform(documents)

# 4. Convert the matrix into a clean DataFrame to see the results
feature_names = vectorizer.get_feature_names_out()
df_bow = pd.DataFrame(vector_matrix.toarray(), columns=feature_names)

print(df_bow)
```
