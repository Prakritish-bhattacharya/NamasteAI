```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Subplots in MaplotLib

```python
# Create a 2 × 2 grid containing 4 separate subplots.
#
# (2, 2, 1) → Selects the 1st position (top-left)
# Plot x^19 against y.
plt.subplot(2, 2, 1)
plt.plot(x**19, y)

# (2, 2, 4) → Selects the 4th position (bottom-right)
# Plot x^38 against y.
plt.subplot(2, 2, 4)
plt.plot(x**38, y)

# (2, 2, 2) → Selects the 2nd position (top-right)
# Plot x against y.
plt.subplot(2, 2, 2)
plt.plot(x, y)

# (2, 2, 3) → Selects the 3rd position (bottom-left)
# Plot y² against x.
plt.subplot(2, 2, 3)
plt.plot(y**2, x)
```

**Object Oriented Way**




`fig.add_axes([left, bottom, width, height])`

```python
# Create a new Figure object.
# A Figure is the overall canvas/container that holds one or more axes.
fig = plt.figure()


# Add the first Axes (plot area) to the Figure.
# [left, bottom, width, height] → controls the position and size
# of the axes relative to the Figure.
#
# left   = 0.1  → 10% from the left
# bottom = 0.1  → 10% from the bottom
# width  = 0.8  → 80% of the Figure width
# height = 0.8  → 80% of the Figure height
axis1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])

# Plot x on the x-axis and y on the y-axis in the first Axes.
axis1.plot(x, y)


# Add a second Axes on top of the first Axes.
# This creates a smaller plot inside the main plot area.
#
# left   = 0.2  → 20% from the left
# bottom = 0.5  → 50% from the bottom
# width  = 0.4  → 40% of the Figure width
# height = 0.3  → 30% of the Figure height
axis2 = fig.add_axes([0.2, 0.5, 0.4, 0.3])

# Plot y on the x-axis and x on the y-axis in the second Axes.
axis2.plot(y, x)
```

# Seaborn


We use distribution plots to visualize the distribution of quantitative data. Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:


1.   dis plot / hist plot
2.   join plot
3.   pair plot
4.   rug plot



```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# load the tip dataset
df = sns.load_dataset("tips")
df
```

```python
df['size'].unique()
```

**dist plot / hist plot**


`displot()` → Distribution of data
`bins=30`   → 30 intervals
`kde=True`  → Smooth distribution curve

```python
# Create a distribution plot for the 'total_bill' column.
# A distribution plot shows how the values are spread across different ranges.
#
# df['total_bill'] → Selects the 'total_bill' column from the DataFrame.
# bins=30 → Divides the data into 30 intervals (bins) and displays
#           the frequency of values in each interval.
# kde=True → Adds a Kernel Density Estimate (KDE) curve to visualize
#            the overall shape of the data distribution.
sns.displot(df['total_bill'], bins=30, kde=True)
```

**Join plot**

```python
# Create a joint plot to visualize the relationship between
# 'total_bill' and 'tip'.
#
# x="total_bill" → Places the 'total_bill' column on the x-axis.
# y="tip"        → Places the 'tip' column on the y-axis.
# data=df        → Uses the DataFrame 'df' as the source of the data.
# kind="scatter" → Displays the relationship using individual
#                  scatter points.
#
# The plot also shows the individual distributions of 'total_bill'
# and 'tip' along the margins.
sns.jointplot(
    x="total_bill",
    y="tip",
    data=df,
    kind="scatter"
)
```

**Pait plot**

```python
# Create a pair plot to visualize pairwise relationships between
# all numerical columns in the DataFrame.
#
# df → The DataFrame containing the data to visualize.
#
# hue="sex" → Groups and colors the data points according to the
#             'sex' column, allowing us to compare the distributions
#             and relationships for different categories.
#
# The diagonal plots show the distribution of each numerical variable,
# while the other plots show relationships between pairs of variables.
sns.pairplot(df, hue="sex")
```

**Rug plot**

```python
# Create a rug plot for the 'total_bill' column.
#
# df['total_bill'] → Selects the 'total_bill' column from the DataFrame.
#
# rugplot() → Displays a small vertical line (tick) for every
#             individual data point along the axis.
#             This helps visualize how the data is distributed
#             and where values are concentrated.
sns.rugplot(df['total_bill'])
```

# Categorical Data Plot

```python
df = sns.load_dataset("tips")
df
```

```python
# Create a count plot to show the number of customers for each gender,
# with the bars further divided based on whether the customer is a smoker.
#
# x=df['sex']:
# Places the 'sex' column on the X-axis.
# It creates separate categories for:
#   - Male
#   - Female
#
# hue=df['smoker']:
# Further divides each gender category according to the 'smoker' column.
# It creates separate bars for:
#   - Yes → customer is a smoker
#   - No  → customer is not a smoker
#
# The height of each bar represents the number of observations
# belonging to that particular combination of gender and smoking status.
#
# This plot helps us compare:
#   1. The number of male and female customers.
#   2. The number of smokers and non-smokers within each gender.
#   3. The relationship between gender and smoking status.
#
# sns.countplot() automatically counts how many rows belong
# to each category, so we do not need to calculate the counts manually.
sns.countplot(x=df['sex'], hue=df['smoker'])
```

```python
# Create a bar plot to compare the total bill amount
# across different gender categories.
#
# x=df['sex']:
# Places the 'sex' column on the X-axis.
# Each category (Male and Female) gets its own bar.
#
# y=df['total_bill']:
# Uses the 'total_bill' column as the numerical variable.
# This is the value that will be aggregated for each gender.
#
# estimator=np.sum:
# Instead of calculating the default statistical estimate (mean),
# calculate the SUM of all 'total_bill' values for each gender.
#
# Therefore, the height of each bar represents:
#
#     Total bills paid by Female customers
#     Total bills paid by Male customers
#
# In simple terms:
#     Gender → Total Bill → Sum
#
# This allows us to compare the overall bill amount
# contributed by each gender category.
sns.barplot(
    x=df['sex'],
    y=df['total_bill'],
    estimator=np.sum
)
```

```python
# Create a box plot to visualize the distribution of tip amounts
# across different days of the week.
#
# x='tip':
# Places the numerical 'tip' column on the X-axis.
# The tip amount is represented along the horizontal axis.
#
# y='day':
# Places the categorical 'day' column on the Y-axis.
# Each day (Thur, Fri, Sat, Sun) gets its own box.
#
# data=df:
# Tells Seaborn to use the DataFrame 'df' as the source of the data.
# Therefore, 'tip' and 'day' are column names from df.
#
# Each box summarizes the distribution of tip values
# for a particular day.
#
# The box plot helps us understand:
#   1. Median tip amount
#   2. Lower and upper quartiles
#   3. Spread of the tip values
#   4. Possible outliers
#   5. Differences in tip distributions across days
sns.boxplot(x='tip', y='day', data=df)
```
