Pandas provides two data structures for manipulating data which are as follows:

`pd.Series()` to convert the array into a Pandas Series.





```python
import pandas as pd
import numpy as np
```

```python
data = np.array(['P','a','n','d','a','s'])
s = pd.Series(data)
print("Pandas series:")
print(s)
```


2.   **Pandas DataFrame:** Pandas DataFrame is a two-dimentional data structure with row and columns. Collection of  multiple series call DataFrame.It is created by loading the datasets from existing storage which can be a SQL database, a CSV file or an Excel file.



```python
import pandas as pd

df = pd.DataFrame()
print(df)
```

```python
list = ['P','a','n','d','a','s']
df = pd.DataFrame(list)
print(df)
```

# Missing Data

```python
import numpy as np
import pandas as pd
```

```python
# Create a DataFrame containing some missing (NaN) values.
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
# Count the number of missing (NaN) values in each column.
df.isna().sum()
```

```python
# Check whether each column contains at least one missing (NaN) value.
df.isna().any()
```

**Remember:** `dropna()` returns a new DataFrame; it does not modify df unless you use `inplace=True`.

```python
# Remove every row that contains at least one missing (NaN) value.
# By default, dropna() works row-wise (axis=0).
df.dropna()
```

```python
df
```

```python
df.dropna(thresh=3)
```

```python
# Replace all missing (NaN) values in the DataFrame with 0.
# By default, fillna() returns a new DataFrame and does not modify the original DataFrame.
df.fillna(0)
```

```python
# Create a dictionary that specifies a different replacement value
# for missing (NaN) values in each column.
values = {
    "A": 0,
    "B": 100,
    "C": 300,
    "D": 400
}

# Replace missing (NaN) values in each column with its corresponding
# value from the 'values' dictionary.
# A → NaN is replaced with 0
# B → NaN is replaced with 100
# C → NaN is replaced with 300
# D → NaN is replaced with 400
#
# fillna() returns a new DataFrame by default and does not modify 'df'.
df.fillna(values)
```

```python
df.fillna(df.mean())
```

# Merging Joining and Concatination

**Merging**

```python
import pandas as pd
import numpy as np
```

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

`inner` → Only matching records from both DataFrames.

```python
# Merge the 'employees' and 'salaries' DataFrames using the common
# 'employee_id' column.
#
# how='inner' → Keeps only the rows where 'employee_id' exists
# in BOTH DataFrames.
#
# on='employee_id' → Specifies the column used to match records
# between the two DataFrames.
#
# pd.merge() returns a new DataFrame and does not modify the
# original 'employees' or 'salaries' DataFrames.
pd.merge(employees, salaries, on='employee_id', how='inner')
```

`outer` → Keeps all records from both DataFrames.

```python
# Merge the 'employees' and 'salaries' DataFrames using the common
# 'employee_id' column.
#
# on='employee_id' → Specifies the column used to match records
# between the two DataFrames.
#
# how='outer' → Keeps ALL records from BOTH DataFrames.
# If an employee_id exists in only one DataFrame, the missing
# values from the other DataFrame will be filled with NaN.
#
# pd.merge() returns a new DataFrame and does not modify the
# original 'employees' or 'salaries' DataFrames.
pd.merge(employees, salaries, on='employee_id', how='outer')
```

# GroupBy

```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}
df = pd.DataFrame(data)
df
```

**Group by Category and Calculate the sum of Sales**

```python
cat = df.groupby('Category')['Sales'].sum()
cat
```

**Group by Store and calculate the sum of Sales**

```python
sto = df.groupby('Store')['Sales'].sum()
sto
```

**Group by multiple columns
Group by Category and Store**

```python
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```

# Pivot Tables

```python
import numpy as np
import pandas as pd

data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
            'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

df = pd.DataFrame(data)

# df['Month'] = df['Date'].dt.month_name()
# df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df, values="Sales", index = "Region", columns = "Product", aggfunc = "median")
```
