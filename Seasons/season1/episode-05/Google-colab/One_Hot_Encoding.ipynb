```python
import pandas as pd

# 1. Create a sample dataset
data = {"Customer": ["Alice", "Bob", "Charlie"], "Favorite_Food": ["Pizza", "Burger", "Tacos"]}
df = pd.DataFrame(data)

# 2. Apply One-Hot Encoding
encoded_df = pd.get_dummies(df, columns=["Favorite_Food"], dtype=int)

print(encoded_df)
```
