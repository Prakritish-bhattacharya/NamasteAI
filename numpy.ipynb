**NumPy** is the foundational open-source library for *scientific computing and data analysis* in Python

# What NumPy Does
* Adds support for large, multi-dimensional arrays and matrices to Python.
* Provides high-performance mathematical, logical, and statistical routines.
* Serves as the core engine for higher-level data tools like pandas, scikit-learn, and TensorFlow

# Why NumPy is Important
* **Speed:**
 Operations run in pre-compiled C code instead of slow, sequential Python loops.
* **Memory Efficiency:** Data is stored in contiguous blocks of memory with uniform types, reducing overhead.
* **Vectorization & Broadcasting:** Enables math operations on entire arrays at once without writing explicit loops.

```python
# import numpy library
import numpy as np
```

**Array Generation function**

`np.arange()`: Generates evenly spaced values within a given interval based on a step size. It works similarly to Python's native `range()`.

```python
# np.arange(start, stop, step)
arr_arange = np.arange(0,10)
print(arr_arange)
# ----------------------------------------
arr2_arange = np.arange(0,10,2)
print(arr2_arange)
```

`np.zeros()`: Creates an array of a specified shape filled entirely with **zeros** (default type is float).

```python
arr_zero = np.zeros((2,3))# ((#2 rows, #3 columns))
arr_zero
```

`np.ones()`: Creates an array filled entirely with **ones**.

```python
arr_ones = np.ones((3,3))
arr_ones
```

`np.linspace()`: Generates an array of evenly spaced numbers over a specified range, but you define the **total number of samples** instead of the step size.

```python
# np.linspace(start, stop, number_of_elements)
arr_linspace = np.linspace(0,10,5)
arr_linspace
```

# Array Attributes

In NumPy, **array attributes** are built-in properties of the `ndarray` object that provide critical details about its structure, data type, and memory layout. You can access these properties directly using dot notation (e.g., `array.shape`) without calling
them like functions.


---


## NumPy Array Attributes

| Attribute | What it Returns | Example Output (for a 2×3 array) |
|---|---|---|
| `.shape` | A tuple showing the size of the array in each dimension (rows, columns). | `(2, 3)` |
| `.ndim` | The number of array dimensions (axes). | `2` |
| `.dtype` | The data type of the elements stored in the array. | `dtype('int64')` |
| `.size` | The total number of elements across all dimensions. | `6` |
| `.itemsize` | The size of a single array element in bytes. | `8` (for a 64-bit integer) |
| `.nbytes` | The total memory consumed by all elements (`size × itemsize`). | `48` (6 elements × 8 bytes) |
