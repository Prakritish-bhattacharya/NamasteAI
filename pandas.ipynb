# Data Structures in Pandas

Pandas provides two data structures for manipulating data which are as follows:



1.   **Pandas Series:** A Pandas Series is a one-dimensional labeled array capable of holding data of any type. It can be created from Python Lists, NumPy arrays, dictionaries from external sources such as CSV files, Excel files and databases.

`pd.Series()` to convert the array into a Pandas Series.





```python
import pandas as pd
import numpy as np
```

```python
data = np.array(['P','a','n','d','a','s'])
s = pd.Series(data)
print("Pandas series:")
print(s)
```


2.   **Pandas DataFrame:** Pandas DataFrame is a two-dimentional data structure with row and columns. Collection of  multiple series call DataFrame.It is created by loading the datasets from existing storage which can be a SQL database, a CSV file or an Excel file.



```python
import pandas as pd

df = pd.DataFrame()
print(df)
```

```python
list = ['P','a','n','d','a','s']
df = pd.DataFrame(list)
print(df)
```

# Let's more explore about Series

```python
import pandas as pd
import numpy as np
```

```python
labels = ['a','b','c']
my_list = [10,20,30] # list
arr = np.array([10,20,30]) # numpy array
d = {'a':10,'b':20,'c':30} # dictionary
```

```python
#  create a series using list
series_list = pd.Series(my_list)
print(series_list)
print("-------------")
# create a series using numpy array
series_numpy_array = pd.Series(arr)
print(series_numpy_array)
print("-------------")
# create a series using dictionary
series_dictionary = pd.Series(d)
print(series_dictionary)
```

```python
# create series and implement index
# take list and index as labels
pd.Series(my_list, index = labels)
```

```python
# take numpy array and index as list
pd.Series(arr, index = my_list)
```

```python
# Series using dictionaries
pd.Series(d)
```

# Let's more explore about DataFrames

```python
#import required libraries
import pandas as pd
import numpy as np

#DataFrames
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
# crate DataFrame
df = pd.DataFrame(data)
```

```python
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]

df = pd.DataFrame(data_list)
df
```

Create custom columns

```python
#  create columns
columns = ["Name","Age","City","Salary"]
df_with_columns = pd.DataFrame(data_list, columns=columns) # pass columns as an Argument--- columns = columns
df_with_columns
```

**Selection and Indexing of Columns**

```python
'''In DataFrame You can't access columns using Indexes...
You need to access it using Column Name'''
df_with_columns["Name"]
df_with_columns[["Name","City"]]
```
