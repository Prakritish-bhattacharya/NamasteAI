```python
from sentence_transformers import SentenceTransformer

# Load a lightweight, pre-trained model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Define sample sentences
sentences = [
    "The puppy played in the garden.",
    "A dog is running outside in the yard.",
    "The stock market experienced a sudden drop.",
]

# Convert sentences to numerical vectors
vectors = model.encode(sentences)

# Print vector shapes and sample values
for i, sentence in enumerate(sentences):
    print(f"Sentence: {sentence}")
    print(f"Vector Shape: {vectors[i].shape}")  # Outputs (384,) -> 384 numbers
    print(f"First 5 dimensions: {vectors[i][:5]}\n")
```
